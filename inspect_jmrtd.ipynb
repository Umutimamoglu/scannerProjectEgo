# Inspect JMRTD Jar

This notebook will inspect the `jmrtd-0.7.31.jar` file and identify the concrete `CardService` implementation or factory method needed for the current `App.java` code.

```python
import zipfile
import os

jar_path = r'C:\Users\Mobiloby\.m2\repository\org\jmrtd\jmrtd\0.7.31\jmrtd-0.7.31.jar'
print('Jar path:', jar_path)

with zipfile.ZipFile(jar_path, 'r') as jar:
    files = [name for name in jar.namelist() if name.endswith('.class') and 'CardService' in name]
    print('CardService-related files:')
    for f in files:
        print(f)
    print('\nTotal files found:', len(files))
```
